```python
import torch
import xitorch as xt
from xitorch.optimize.rootfinder import equilibrium
from torchviz import make_dot
```

```python
torch.set_default_dtype(torch.float64)
torch.set_printoptions(precision=16)
```

```python
def func(z, x, B):
    return torch.sqrt(z + x + B)
```

```python
X = torch.tensor([[13.1, 12.0], [1, 12.0]], requires_grad=True)
print(f"Input X: {X}")
```

```python
y0 = torch.zeros_like(X)
y0
```

```python
class b_test:
    def __init__(self, val):
        self.val = val
```

```python
B = torch.tensor([1.0], requires_grad = True)
```

```python
yequil = equilibrium(func, y0, params=(B, X))
```

```python
yequil
```

```python
make_dot(yequil).render("xitorch_test", format="png")
```
