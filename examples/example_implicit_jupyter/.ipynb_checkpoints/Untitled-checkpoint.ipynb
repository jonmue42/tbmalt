```python
def implicit(fn, z, *args, **kwargs):
    #final = fn(z, x)
    #create a second graph for derivative calculation regarding the final value
    z_2 = z.detach()
    z_2.requires_grad = True
    #x_2 = x.detach()
    final_detached = fn(z_2, *args, **kwargs)
    make_dot(final_detached).render("final_detached2", format="png")
    #final_detached_sum = final_detached.sum()
    implicit_contrib = torch.autograd.grad(final_detached, z_2, torch.ones_like(final_detached), retain_graph = True)[0]#, allow_unused = True)[0]
    print(implicit_contrib)
    final_detached.register_hook(lambda grad: grad/(1-implicit_contrib))
    return final_detached
```

```python
class implicit_fixedpoint(torch.autograd.Function):
    @staticmethod
    def forward(x, tol, max_iter):
        z = torch.zeros_like(x)
        iterations = 0

        while iterations < max_iter:
            z_next = torch.sqrt(z + x)
            g = z - z_next
            err = torch.norm(g)
            if err < tol:
                break
            z = z_next
            iterations += 1
        
        z = torch.sqrt(z + x)
        return z

    @staticmethod
    def setup_context(ctx, inputs, outputs):
        x, tol, max_iter = inputs
        z = outputs
        ctx.save_for_backward(x, z)

    @staticmethod
    def backward(ctx, grad_output):
        x, z = ctx.saved_tensors
        grad_x = None
        grad_x = 20*grad_output

        with torch.enable_grad():
            z_back = torch.sqrt(z + x)
            implicit_contrib = torch.autograd.grad(z_back, z, torch.ones_like(z))[0]
            print(implicit_contrib)

        grad_x =  grad_output * (1/ (1 - implicit_contrib) - 1)
        
        return grad_x, None, None
```

```python
def _mulliken(
        rho: torch.Tensor, S: torch.Tensor, orbs = None,
        resolution: Optional[Literal['atom', 'shell', 'orbital']] = None
) -> torch.Tensor:
    r"""Mulliken population analysis.

    By default, orbital resolved populations are returned, however, passing the
    associated orbs instance will produce atom or shell resolved populations
    depending on the orbs instance's `shell_resolved` attribute (the behavior
    of which can be overridden via the ``resolution`` argument).

    Arguments:
        rho: density matrix.
        S: overlap matrix.
        orbs: a `OrbitalInfo` instance may be specified to enable atom/shell
            resolved populations. If omitted, populations will be orbital
            resolved. [DEFAULT=None]
        resolution: can be specified to override the degree of resolution
            defined by the ``orbs`` instance, available options are:

                - "atom": atom resolved
                - "shell": shell resolved
                - "orbital": orbital resolved

            If unspecified, this will default to the resolution defined by the
            `orbs.shell_resolved` attribute. This is only valid when ``orbs``
            is also specified. [DEFAULT=None]

    Returns:
        q: mulliken populations.

    Raises:
        TypeError: if ``resolution`` is specified in absence of ``orbs``.

    """

    if resolution is not None and orbs is None:
        raise TypeError(
            '"resolution" overrides default behaviour associated with the '
            '"orbs" object\'s "shell_resolved" attribute. Thus it cannot be '
            'specified in absence of the "orbs" argument.')

    q = (rho * S).sum(-1)  # Calculate the per-orbital Mulliken populations
    if orbs is not None:  # Resolve to per-shell/atom if instructed to
        if resolution is None:
            # TODO: Change orbs to have a res_matrix_shape property.
            size, ind = orbs.res_matrix_shape, orbs.on_res
        elif resolution == 'atom':
            size, ind = orbs.atomic_matrix_shape, orbs.on_atoms
        elif resolution == 'shell':
            size, ind = orbs.shell_matrix_shape, orbs.on_shells
        else:
            raise NotImplementedError("Unknown resolution")

        q = torch.zeros(size[:-1], device=rho.device, dtype=rho.dtype
                        ).scatter_add_(-1, ind.clamp(min=0), q)

    return q
```

```python
def finite_diff(func, inp, delta):
    finitediff = 0
    for indx in range(len(inp)):
        new_inp = inp.clone()
        new_inp[indx] += delta
        print(new_inp)
        finitediff += (func(new_inp) - func(inp))/delta
    return finitediff
```

```python
def _scc_cycle(q_in: torch.Tensor) -> torch.Tensor:
    """Perform a single self-consistent charge cycle.

    This method performs a single Self-Consistent Charge cycle (SCC). Using
    ``q_in`` as the initial guess during the first cycle and as the mixed
    charges in all subsequent cycles.

    Arguments:
        q_in: Input charges are provided via the ``q_in`` argument. This
            can be viewed as the initial guess for the first cycle and as
            the mixed charges from the previous step in all subsequent
            cycles.

    Returns:
        q_out: New updated charges, as computed following the SCC step.

    Notes:
        It is important to note that this method is intended to modify and
        update the class and its attributes during the SCC cycle. The newly
        computed charges are only returned to facilitate ease of use.

        The charges ``q_in`` and ``q_out`` may be either shell or atom
        resolved, but must match up with that as defined by the orbs
        attribute `shell_resolved`.
    """

    # Construct the shift matrix
    def shift_construction(q_in):
        shifts = torch.einsum(
            '...i,...ij->...j', q_in - q_zero_res, gamma)
        #print("SHIFTS1")
        #print(shifts)
        shifts = prepeat_interleave(shifts, orbs.orbs_per_res)
        #print("SHIFTS2")
        #print(shifts)
        shifts = (shifts[..., None] + shifts[..., None, :])
        #print("SHIFTS3")
        #print(shifts)
        return shifts
    shifts = shift_construction(q_in)
    #print("SHIFTS")
    #print(shifts)
    grad_shift = torch.autograd.grad(shifts, q_in, torch.ones_like(shifts), retain_graph = True)
    print("Grad shift")
    print(grad_shift[0].sum())
    finitediff_shift = finite_diff(shift_construction, q_in, 1e-9)
    print("finite diff shift")
    print(finitediff_shift.sum())

    # Compute the second order Hamiltonian matrix
    def hamiltonian_construction(q_in):
        shifts = shift_construction(q_in)
        _hamiltonian = core_hamiltonian + .5 * overlap * shifts
        return _hamiltonian
    _hamiltonian = hamiltonian_construction(q_in)
    grad_hamiltonian = torch.autograd.grad(_hamiltonian, q_in, torch.ones_like(_hamiltonian), retain_graph = True)
    print("Grad hamiltonian")
    print(grad_hamiltonian[0].sum())
    print("finite diff hamiltonian")
    finitediff_hamiltonian = finite_diff(hamiltonian_construction, q_in, 1e-9)
    print(finitediff_hamiltonian.sum())

    

    # Obtain the eigen-values/vectors via an eigen decomposition
    eig_values, eig_vectors = eighb(
        _hamiltonian, overlap, **_solver_settings)
    grad_eigvalues = torch.autograd.grad(eig_values, q_in, torch.ones_like(eig_values), retain_graph = True)
    print(grad_eigvalues)
    grad_eigvectors = torch.autograd.grad(eig_vectors, q_in, torch.ones_like(eig_vectors), retain_graph = True)
    print(grad_eigvectors)


    # Scaled occupancy values
    s_occs = torch.einsum(
        '...i,...ji->...ji', torch.sqrt(occupancy), eig_vectors)
    grad_soccs = torch.autograd.grad(s_occs, q_in, torch.ones_like(s_occs), retain_graph = True)
    print(grad_soccs)


    # Density matrix
    rho = s_occs @ s_occs.transpose(-1, -2).conj()
    grad_rho = torch.autograd.grad(rho, q_in, torch.ones_like(rho), retain_graph = True)
    print(rho)

    # Compute and return the new
    return _mulliken(rho, overlap, orbs)
    
```
